```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "My ML API is working!"}

print("API created successfully!")
```

```python
result = home()

print(result)
```

```python
def predict_message(message):
    if "win" in message.lower() or "prize" in message.lower():
        return "Spam"
    else:
        return "Ham"

print(predict_message("You won a prize"))
print(predict_message("Hello, how are you?"))
```

```python
from pydantic import BaseModel

class Message(BaseModel):
    message: str

@app.post("/predict")
def predict(data: Message):
    result = predict_message(data.message)

    return {
        "message": data.message,
        "prediction": result
    }

print("Prediction API created successfully!")
```

```python
test1 = predict(Message(message="You won a free prize"))
test2 = predict(Message(message="Hello, how are you?"))

print(test1)
print(test2)
```
